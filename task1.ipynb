# Task 1 – Data Immersion & Wrangling

## ApexPlanet Sales Dataset

**Objective:** Understand, clean, transform, and prepare the sales dataset for further analysis.

This notebook uses the dataset provided for the ApexPlanet Task 1.


## 1. Import Libraries

Pandas and NumPy are used for data loading, exploration, cleaning, and transformation.


```python
import pandas as pd
import numpy as np
```

## 2. Load the Dataset

The Excel workbook contains the `Sales_Dataset` sheet.


```python
file_path = "ApexPlanet_DataAnalytics_Dataset (1).xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="Sales_Dataset"
)

print("Dataset loaded successfully!")
df.head()
```

## 3. Data Exploration

We inspect the dataset shape, columns, data types, and statistical summary.


```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
df.describe(include="all").T
```

## 4. Missing Value Analysis

We check each column for missing values before cleaning.


```python
missing_values = df.isnull().sum()
display(missing_values.to_frame("Missing_Count"))
```

## 5. Duplicate Record Check

We check whether the dataset contains duplicate rows.


```python
print("Number of duplicate rows:", df.duplicated().sum())
```

## 6. Data Cleaning

Cleaning steps:
- Convert `Order_Date` to datetime.
- Fill missing `Age` values using the median.
- Fill missing `City` values using the most frequent city.
- Remove extra spaces from text fields.
- Standardize text capitalization.
- Remove duplicate records.


```python
# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")

# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing City with mode
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Clean text columns
for col in ["Gender", "City", "Product", "Category"]:
    df[col] = df[col].astype("string").str.strip()

# Standardize capitalization
df["Gender"] = df["Gender"].str.title()
df["City"] = df["City"].str.title()
df["Product"] = df["Product"].str.title()
df["Category"] = df["Category"].str.title()

# Remove duplicate rows
df = df.drop_duplicates()

print("Cleaning completed!")
```

## 7. Final Data Validation

We verify missing values, duplicate records, data types, and the final dataset shape after cleaning.


```python
print("Missing values after cleaning:")
display(df.isnull().sum().to_frame("Missing_Count"))

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())

print("\nFinal dataset shape:")
print(df.shape)

print("\nData types:")
display(df.dtypes.to_frame("Data_Type"))
```

## 8. Export Cleaned Dataset

The cleaned dataset is exported as a CSV file for further analysis.


```python
output_file = "ApexPlanet_Cleaned_Sales_Dataset.csv"

df.to_csv(output_file, index=False)

print(f"Cleaned CSV file created successfully: {output_file}")
```
